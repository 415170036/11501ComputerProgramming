## 第三部分：CPython 如何實作這些型別？

除了知道 Python 使用哪些型別表示數學物件之外，

我們也希望了解：

> Python 物件在 CPython 中是如何實作的？

請查閱資料，完成下表。

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|
| 自然數 | `int` | PyLongObject | 可表示任意大小整數 |
| 整數 | `int` | PyLongObject | 支援正負整數與大數運算 |
| 實數 | `float` | PyFloatObject | 採用雙精度浮點數（IEEE 754） |
| 複數 | `complex` | PyComplexObject | 同時儲存實部與虛部 |
| 集合 | `set` | PySetObject | 元素不重複，支援快速查找 |
| 命題真假 | `bool` | PyBoolObject | 僅有 True 與 False |
| 串列 | `list` | PyListObject | 可變動、有順序、可存放不同型別資料 |

## 第四部分：深入探討各種型別

請針對第三部分中的每一種型別回答：

1. 它是否為 Python 的內建型別？
2. 對應的 CPython 物件結構名稱為何？
3. 它如何儲存資料？
4. 是否有固定大小限制？
5. 若資料規模持續增加，如何擴充儲存空間？
(1)int：1.是 2.PyLongObject 3.以多個數字區塊（digits）儲存整數，而不是固定 32 位元或 64 位元 4.沒有 5.CPython 會自動配置更多記憶體來儲存額外的數字區塊
(2)float：1.是 2.PyFloatObject 3.使用 IEEE 754 雙精度浮點數（double） 4.有 5.無法擴充
(3)complex：1.是 2.PyComplexObject 3.包含兩個 double：實部(real)、虛部(imag) 4.有 5.無法動態擴充
(4)bool：1.是 2.PyBoolObject 3.只有兩個物件：True、False 4.有 5.不需要擴充
(5)set：1.是 2.PySetObject 3.使用Hash Table（雜湊表） 4.沒有 5.當負載過高時：用Resize建立更大的 Hash Table，並重新配置元素
(6)list：1.是 2.PyListObject 3.使用動態陣列（Dynamic Array）儲存物件參考（Reference） 4.沒有 5.CPython 採用Over-allocation策略

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？因為 Python 3 的 int 採用了 任意精度整數（Arbitrary Precision Integer） 的設計
2. Python 的 `int` 是否具有固定的位元數限制？沒有固定的位元數限制
3. 什麼是 Arbitrary Precision（任意精度）？整數能使用任意多的位數進行表示，而不是限制在 32 位元或 64 位元
4. CPython 使用什麼物件來實作 `int`？CPython 底層使用PyLongObject來實作 Python 的整數型別
5. Python 整數大小真正的限制是什麼？(1) 電腦記憶體大小(2) 作業系統限制(3) 執行時間限制

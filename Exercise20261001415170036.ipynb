5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。

1. 計算過程：
0.1 × 2 = 0.2  → 0
0.2 × 2 = 0.4  → 0
0.4 × 2 = 0.8  → 0
0.8 × 2 = 1.6  → 1

取小數部分 0.6 繼續

0.6 × 2 = 1.2  → 1
0.2 × 2 = 0.4  → 0
0.4 × 2 = 0.8  → 0
0.8 × 2 = 1.6  → 1

又回到 0.6 ...

因此：
0.1₁₀
=
0.0001100110011001100110011...₂

寫成循環表示：
0.1₁₀ = 0.0001(1001)₂

其中：
1001
會一直重複下去。

觀察到的現象：
0.1 在十進位是有限小數：
0.1

但是在二進位卻變成：
0.000110011001100110011...
是一個無限循環小數。

因此電腦無法用有限位元完整儲存它。

2. Python 的 float 採用：
IEEE 754 Double Precision

也就是：
64 bits
格式。

由於位元數有限，因此：
0.000110011001100110011...
不可能全部存下來。

Python 只能：
截斷
+
四捨六入五成雙
取得最接近的近似值。

例如：
0.1

實際儲存的是：
0.10000000000000000555...
而不是數學上的精確 0.1。

4. 執行：
print(0.1+0.2)

結果：
0.30000000000000004

原因：
0.1
與
0.2
本身就是近似值。

實際上：
0.10000000000000000555...
+
0.20000000000000001110...

得到：
0.30000000000000004440...

所以顯示：
0.30000000000000004

方法一：格式化輸出
print(f"{0.1+0.2:.1f}")

結果：
0.3

方法二：使用 round()
print(round(0.1+0.2, 1))

結果：
0.3

方法三：比較時使用 isclose()

不要寫：
0.1 + 0.2 == 0.3

應寫：
import math
 
print(math.isclose(0.1+0.2, 0.3))

結果：
True

方法四：使用 Decimal（金融常用）
from decimal import Decimal
 
print(
Decimal("0.1")
+ Decimal("0.2")
)

結果：
0.3
